### Análise dos Dados do IDEB (Ensino Médio - 2023)

Este notebook tem como objetivo processar e analisar os dados do IDEB (Índice de Desenvolvimento da Educação Básica) para escolas de ensino médio no Brasil. As etapas incluem:

* **Carregamento dos dados**: Importação do arquivo CSV com informações do IDEB.
* **Limpeza e organização**: Ajuste dos nomes das colunas e preparação dos dados.
* **Filtragem por estado**: Seleção das escolas do estado de São Paulo (SP) para análise.
* **Visualização inicial**: Exibição de amostras dos dados processados.

O foco está em estruturar os dados de forma eficiente para análises futuras, mantendo a transparência e clareza no processo de manipulação.

```python
# Importa a biblioteca pandas para manipulação de dados
import pandas as pd
```

```python
# Carrega as linhas de interesse do arquivo CSV
data_ideb = pd.read_csv(
    'divulgacao_ensino_medio_escolas_2023.csv',  # Arquivo de dados
    encoding='ISO-8859-1',  # Codificação do arquivo
    sep=';',  # Separador de colunas
    low_memory=False,  # Reduz avisos sobre tipos mistos
    skiprows=9,  # Ignora as 9 primeiras linhas
    nrows=21587 - 9  # Lê até a linha 21587
)
```

```python
# Exibe as primeiras 5 linhas do DataFrame para visualização
data_ideb.head()
```

```python
# Limpar os nomes das colunas
data_ideb.columns = (
    data_ideb.columns
    .str.replace("\n", " ")  # Substituir quebras de linha por espaços
    .str.replace(r"\s+", " ", regex=True)  # Substituir múltiplos espaços por um único espaço
    .str.strip()  # Remover espaços extras no início e fim
)

# Exibir os novos nomes das colunas
data_ideb.columns
```

```python
# Filtrar apenas as colunas desejadas
colunas_desejadas = ["SG_UF", "CO_MUNICIPIO", "NO_MUNICIPIO", "ID_ESCOLA", "VL_OBSERVADO_2023"]
data_ideb_filtrados = data_ideb[colunas_desejadas].copy()
```

```python
# Filtra os dados para incluir apenas registros do estado de São Paulo (SP)
data_ideb_filtrados_sp = data_ideb_filtrados[data_ideb_filtrados['SG_UF'] == 'SP']
```

```python
# Exibe as primeiras 5 linhas dos dados filtrados para São Paulo
data_ideb_filtrados_sp.head()
```

```python
# Exportar o DataFrame para um arquivo CSV
data_ideb_filtrados_sp.to_csv("data_ideb_filtrados_sp.csv", index=False, encoding="utf-8")

# Mensagem de confirmação
print("Arquivo 'data_ideb_filtrados_sp.csv' exportado com sucesso!")
```
